# Mini Tokenizer for a Pseudo Language

Objective:
To design a mini tokenizer that identifies different token types in a custom pseudo-language using Python regular expressions.

Import Libraries

```python
import re
```

Define Tokens and Regular Expressions

```python
token_patterns = [
    ("KEYWORD", r"\b(?:let|show|if|else|repeat|while|true|false)\b"),
    ("IDENTIFIER", r"[a-zA-Z_][a-zA-Z0-9_]*"),
    ("FLOAT", r"\d+\.\d+"),
    ("INTEGER", r"\d+"),
    ("STRING", r'"[^"\n]*"'),
    ("OPERATOR", r"==|!=|<=|>=|\+|-|\*|/|=|<|>"),
    ("SYMBOL", r"[();{},]"),
    ("WHITESPACE", r"\s+"),
]
```

Create the Tokenizer Function

```python
def tokenize(code):
    tokens = []
    position = 0

    while position < len(code):
        match = None

        for token_type, pattern in token_patterns:
            match = re.match(pattern, code[position:])

            if match:
                value = match.group(0)

                if token_type != "WHITESPACE":
                    tokens.append((token_type, value))

                position += len(value)
                break

        if match is None:
            print(f"Unknown character '{code[position]}' at position {position}")
            position += 1

    return tokens
```

Enter a MiniLang Program

```python
code = """
let score = 95
let name = "Ria"
let average = 87.5

if (score >= 90) {
    show("Excellent")
} else {
    show("Keep practicing")
}
"""
```

Generate Tokens

```python
result = tokenize(code)

print("Generated Tokens:\n")

for token_type, value in result:
    print(f"{token_type:<12} : {value}")
```

Display Token Counts

```python
from collections import Counter

counts = Counter(token_type for token_type, value in result)

print("Token Counts:\n")

for token_type, count in counts.items():
    print(f"{token_type}: {count}")

print("Total Tokens:", len(result))
```

Test Another Input

```python
test_code = """
let age = 21
if (age >= 18) {
    show("Adult")
}
"""

test_result = tokenize(test_code)

print("Input Program:\n", test_code)
print("\nGenerated Tokens:")

for token_type, value in test_result:
    print(f"{token_type:<12} : {value}")
```

Test Invalid Characters

```python
test_code = "let value = 10 @ 5"

print("Testing invalid input:\n")

test_result = tokenize(test_code)

print("\nRecognized Tokens:")

for token_type, value in test_result:
    print(f"{token_type:<12} : {value}")
```

Token Definitions and Regular Expressions

# Token Definitions

| Token | Regular Expression / Rule | Example |
|---|---|---|
| KEYWORD | Predefined words | let, if, else, show |
| IDENTIFIER | [a-zA-Z_][a-zA-Z0-9_]* | score, average |
| INTEGER | \d+ | 10, 95 |
| FLOAT | \d+\.\d+ | 87.5 |
| STRING | Text inside double quotes | "Hello" |
| OPERATOR | Arithmetic, relational, assignment | +, >=, = |
| SYMBOL | Punctuation symbols | (, ), {, }, ; |
| WHITESPACE | \s+ | spaces, tabs, newlines |

Whitespace is ignored, while unknown characters generate warnings.

Algorithm

# Algorithm

1. Read the pseudo-language source code.
2. Define token types and their regular expressions.
3. Initialize the current position to zero.
4. Match token patterns against the remaining input.
5. Identify keywords, identifiers, numbers, strings, operators, and symbols.
6. Ignore whitespace.
7. Store each recognized token with its type and value.
8. If no pattern matches, report an unknown character and advance the position.
9. Continue until the complete input is scanned.
10. Display the generated tokens and their counts.
