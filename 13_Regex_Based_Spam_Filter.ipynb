# Spam Filter Using Regular Expressions

Objective: To design a spam filter using Python regular expressions to identify suspicious words, promotional offers, and spam-like messages.

Import Libraries

```python
import re
```

Define Spam Patterns

```python
spam_patterns = [
    r"\bwin\b",
    r"\bfree\b",
    r"\bprize\b",
    r"\bcongratulations\b",
    r"\bclaim now\b",
    r"\bclick here\b",
    r"\blottery\b",
    r"\b\d+% off\b",
    r"\bearn money\b",
    r"\bact now\b"
]
```

Define the Spam Detection Function

```python
def detect_spam(message):
    for pattern in spam_patterns:
        if re.search(pattern, message, re.IGNORECASE):
            return True, pattern
    return False, None
```

Enter a Message

```python
message = input("Enter a message to check: ")
print("Message:", message)
```

Check Whether the Message Is Spam

```python
is_spam, matched_pattern = detect_spam(message)

if is_spam:
    print("Result: SPAM")
    print("Matched Pattern:", matched_pattern)
else:
    print("Result: NOT SPAM")
```

Test Multiple Messages

```python
test_messages = [
    "Congratulations! You won a free prize.",
    "Click here to claim now",
    "Get 50% off today",
    "Hey, are we meeting for lunch?",
    "The lecture starts at 10 AM."
]

for msg in test_messages:
    is_spam, pattern = detect_spam(msg)
    result = "SPAM" if is_spam else "NOT SPAM"
    print(f"{msg} --> {result}")
```

Display Spam Patterns

| Pattern | Purpose |
|---|---|
| `\bwin\b` | Detects the word "win" |
| `\bfree\b` | Detects the word "free" |
| `\bprize\b` | Detects prize-related messages |
| `\bclaim now\b` | Detects claim requests |
| `\bclick here\b` | Detects suspicious links |
| `\b\d+% off\b` | Detects percentage discounts |
| `\bearn money\b` | Detects money-making offers |

Algorithm

1. Define regular expressions for common spam phrases.
2. Accept a message from the user.
3. Compare the message against each pattern using `re.search()`.
4. Ignore case differences during matching.
5. If any pattern matches, classify the message as SPAM.
6. Otherwise, classify it as NOT SPAM.
7. Test the filter with multiple messages.
