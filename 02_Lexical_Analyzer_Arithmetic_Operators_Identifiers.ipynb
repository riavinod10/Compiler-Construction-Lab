```python
%%writefile lexer.py

import ply.lex as lex

# List of tokens
tokens = (
    'INTEGER',
    'FLOAT',
    'IDENTIFIER',
    'PLUS',
    'MINUS',
    'MULTIPLY',
    'ASSIGN',
    'EQUAL'
)

# Regular expressions for operators
t_PLUS = r'\+'
t_MINUS = r'-'
t_MULTIPLY = r'\*'
t_EQUAL = r'=='
t_ASSIGN = r'='

# Floating-point numbers
def t_FLOAT(t):
    r'\d+\.\d+'
    t.value = float(t.value)
    return t

# Integers
def t_INTEGER(t):
    r'\d+'
    t.value = int(t.value)
    return t

# Identifiers
def t_IDENTIFIER(t):
    r'[a-zA-Z_][a-zA-Z0-9_]*'
    return t

# Ignore spaces and tabs
t_ignore = " \t"

# Count new lines
def t_newline(t):
    r'\n+'
    t.lexer.lineno += len(t.value)

# Error handling
def t_error(t):
    print("Illegal character:", t.value[0])
    t.lexer.skip(1)

# Build the lexer
lexer = lex.lex()
```
