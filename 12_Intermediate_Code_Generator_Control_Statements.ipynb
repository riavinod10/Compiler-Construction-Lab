# Intermediate Code Generation for Control Structures

Objective: To generate intermediate code for control structures using three-address code.

Define the Intermediate Code Generator

```python
def generate_while_code(condition, statements):
    code = []
    code.append("L1:")

    code.append(f"if not ({condition}) goto L2")

    for statement in statements:
        code.append(statement)

    code.append("goto L1")
    code.append("L2:")

    return code
```

Enter the While Loop

```python
condition = input("Enter the loop condition: ")
n = int(input("Enter number of statements: "))

statements = []
for i in range(n):
    statements.append(input(f"Enter statement {i+1}: "))
```

Generate Intermediate Code

```python
intermediate_code = generate_while_code(condition, statements)

print("Generated Intermediate Code:\n")
for instruction in intermediate_code:
    print(instruction)
```

Test Example

```python
condition = "i < 5"
statements = ["sum = sum + i", "i = i + 1"]

print("While loop intermediate code:\n")
for instruction in generate_while_code(condition, statements):
    print(instruction)
```

Algorithm

1. Read the loop condition and statements.
2. Generate a label for the beginning of the loop.
3. Check the condition and jump to the exit label if it is false.
4. Execute the loop statements.
5. Jump back to the beginning of the loop.
6. Generate the exit label and display the intermediate code.
