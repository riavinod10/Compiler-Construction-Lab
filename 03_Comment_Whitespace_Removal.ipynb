# **A1. Write a program in Python to remove comments in a C program.**

Input C Program

```python
# Input C Program

lines = [
    "int main()",
    "{",
    "// test program",
    'printf("Hello World");',
    "/* block",
    "comment */",
    "int a = 10;",
    "return 0;",
    "}"
]
```

Function to Remove Comments

```python
def remove_comments(lines):
    output = []
    line_numbers = []

    inside_block = False

    for line_no, line in enumerate(lines, start=1):

        new_line = ""
        i = 0

        while i < len(line):

            # Ignore characters inside block comments
            if inside_block:
                if i + 1 < len(line) and line[i] == '*' and line[i + 1] == '/':
                    inside_block = False
                    i += 2
                else:
                    i += 1

            # Beginning of block comment
            elif i + 1 < len(line) and line[i] == '/' and line[i + 1] == '*':
                inside_block = True
                i += 2

            # Single-line comment
            elif i + 1 < len(line) and line[i] == '/' and line[i + 1] == '/':
                break

            else:
                new_line += line[i]
                i += 1

        if new_line.strip():
            output.append(new_line.rstrip())
            line_numbers.append(line_no)

    return output, line_numbers
```

Remove Comments

```python
clean_program, line_numbers = remove_comments(lines)
```

Display Output

```python
print("C Program Without Comments:\n")

for line in clean_program:
    print(line)

print("\nArray of Original Line Numbers:")

print(line_numbers)
```

# **Practice Question 2) Add Error Handling to the Implementation**

Input C Program

```python
lines = [
    "int main()",
    "{",
    "// sample program",
    "printf(\"Hello World\");",
    "}"
]
```

Remove Comments with Error Handling

```python
try:
    clean_program, line_numbers = remove_comments(lines)

    print("Program Without Comments:\n")

    for line in clean_program:
        print(line)

    print("\nOriginal Line Numbers:")
    print(line_numbers)

except NameError:
    print("Error: Input program not found.")

except Exception as e:
    print("Unexpected Error:", e)
```
