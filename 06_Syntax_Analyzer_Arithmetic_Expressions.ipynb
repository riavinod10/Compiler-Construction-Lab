Define Lexer

```python
import sys
import ply.lex as lex

# Fix for Google Colab
sys.modules['__main__'].__file__ = 'lexer.py'

tokens = (
    'NUMBER',
    'PLUS',
    'MINUS',
    'TIMES',
    'DIVIDE'
)

t_PLUS = r'\+'
t_MINUS = r'-'
t_TIMES = r'\*'
t_DIVIDE = r'/'

def t_NUMBER(t):
    r'\d+'
    t.value = int(t.value)
    return t

t_ignore = ' \t'

def t_newline(t):
    r'\n+'
    t.lexer.lineno += len(t.value)

def t_error(t):
    print(f"Illegal character '{t.value[0]}' at line {t.lineno}, position {t.lexpos}")
    t.lexer.skip(1)

lexer = lex.lex()
```

Define Parser

```python
import ply.yacc as yacc

precedence = (
    ('left', 'PLUS', 'MINUS'),
    ('left', 'TIMES', 'DIVIDE'),
)

def p_expression_number(p):
    'expression : NUMBER'
    p[0] = p[1]

def p_expression_binop(p):
    '''expression : expression PLUS expression
                  | expression MINUS expression
                  | expression TIMES expression
                  | expression DIVIDE expression'''

    if p[2] == '+':
        p[0] = p[1] + p[3]
    elif p[2] == '-':
        p[0] = p[1] - p[3]
    elif p[2] == '*':
        p[0] = p[1] * p[3]
    elif p[2] == '/':
        p[0] = p[1] / p[3]

def p_error(p):
    if p is None:
        print("Syntax error at EOF")
    else:
        print(f"Syntax error at '{p.value}'")

parser = yacc.yacc(write_tables=False, debug=False)
```

Test 2+3

```python
expression = "2+3"

lexer.lineno = 1
result = parser.parse(expression, lexer=lexer)

if result is not None:
    print("Result:", result)
```

# Algorithm

1. Read an arithmetic expression from the user.
2. The lexical analyzer scans the expression.
3. Numbers and operators are converted into tokens.
4. Illegal characters are reported as errors.
5. The parser receives the tokens from the lexer.
6. The parser checks the expression using grammar rules.
7. If the expression is invalid, a syntax error is displayed.
8. If the expression ends unexpectedly, "Syntax error at EOF" is displayed.
9. If the expression is valid, the semantic actions calculate the result.
10. Display the result.

# Grammar

expression → expression + expression
           | expression - expression
           | expression * expression
           | expression / expression
           | NUMBER

Operator precedence:
* and / have higher precedence than + and -.

# Working Example

Input:
2 + 3 * 4

LEX converts the input into tokens:

NUMBER PLUS NUMBER TIMES NUMBER

YACC checks these tokens using the grammar.

Since * has higher precedence:

3 * 4 = 12

Then:

2 + 12 = 14

Output:
Result: 14
