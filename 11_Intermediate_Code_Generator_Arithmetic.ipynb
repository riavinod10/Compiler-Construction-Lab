# Intermediate Code Generation for Arithmetic Expressions

Objective: To generate three-address intermediate code for arithmetic expressions using the operators +, -, * and /.

Import Libraries

```python
import re
```

Define the Intermediate Code Generator

```python
def generate_intermediate_code(expression):
    tokens = re.findall(r'[a-zA-Z_]\w*|\d+|[+\-*/()]', expression)
    operators = []
    operands = []
    precedence = {'+': 1, '-': 1, '*': 2, '/': 2}
    code = []
    temp_count = 1

    def generate():
        nonlocal temp_count
        operator = operators.pop()
        right = operands.pop()
        left = operands.pop()
        temp = f't{temp_count}'
        temp_count += 1
        code.append(f'{temp} = {left} {operator} {right}')
        operands.append(temp)

    for token in tokens:
        if token.isalnum() or '_' in token:
            operands.append(token)

        elif token == '(':
            operators.append(token)

        elif token == ')':
            while operators and operators[-1] != '(':
                generate()
            if operators:
                operators.pop()

        else:
            while (operators and operators[-1] != '('
                   and precedence[operators[-1]] >= precedence[token]):
                generate()
            operators.append(token)

    while operators:
        generate()

    print("Intermediate Code:")
    for instruction in code:
        print(instruction)

    print("Final Result:", operands[-1])
```

Enter an Arithmetic Expression

```python
expression = input("Enter an arithmetic expression: ")
print("Expression:", expression)
```

Generate Intermediate Code

```python
generate_intermediate_code(expression)
```

Test Another Expression

```python
expression = "a + b * c - d / e"
print("Expression:", expression)
generate_intermediate_code(expression)
```

Algorithm

1. Read the arithmetic expression.
2. Identify operands, operators, and parentheses.
3. Apply operator precedence, giving * and / higher priority than + and -.
4. Generate a temporary variable for each operation.
5. Store each operation as a three-address instruction.
6. Display the generated intermediate code and final temporary variable.
