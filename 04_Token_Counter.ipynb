# Lexical Analyzer in Python

## Objective
To read a C program and perform lexical analysis using Python regular expressions.

The program identifies and counts:
- Keywords
- Identifiers
- Operators
- Numbers
- Total Tokens

Import Required Library

```python
import re
```

Define Keywords and Operators

```python
# C keywords
keywords = {
    "if", "else", "while", "for",
    "int", "float", "char", "return",
    "void", "main", "include",
    "struct", "typedef", "enum"
}

# Operators and delimiters
operators_set = {
    "+", "-", "*", "/", "%", "=",
    "==", "!=", "<", ">", "<=", ">=",
    "&&", "||", "!", "&", "|", "^",
    "++", "--",
    "+=", "-=", "*=", "/=", "%=",
    "&=", "|=", "^=",
    "->", "~", "?", ":",
    ";", ",", "(", ")", "{", "}",
    "[", "]"
}
```

Define Token Patterns

```python
token_patterns_ordered = [

    # Comments
    ("COMMENT", r'//[^\n]*|/\*[\s\S]*?\*/'),

    # Whitespace
    ("WHITESPACE", r'\s+'),

    # Numbers - float before integer
    ("NUMBER", r'\d+\.\d+|\d+'),

    # Multi-character operators before single-character operators
    ("OPERATOR", r'==|!=|<=|>=|\+=|-=|\*=|/=|%=|&=|\|=|\^=|->|\+\+|--|&&|\|\|'),

    # Identifiers
    ("IDENTIFIER", r'[A-Za-z_][A-Za-z0-9_]*'),

    # Single-character operators and delimiters
    ("OPERATOR", r'[+\-*/%=<>!&|^~?:;,\(\)\{\}\[\]]')
]
```

Define the Lexical Analyzer

```python
def analyze_c(source_code):
    counts = {
        "Keywords": 0,
        "Identifiers": 0,
        "Operators": 0,
        "Numbers": 0,
        "Total Tokens": 0
    }

    tokens = []

    pos = 0
    length = len(source_code)

    while pos < length:
        match = None

        for token_type, pattern in token_patterns_ordered:
            match = re.match(pattern, source_code[pos:])

            if match:
                value = match.group(0)

                # Ignore comments and whitespace
                if token_type == "COMMENT" or token_type == "WHITESPACE":
                    pos += len(value)
                    break

                # Identifier: check whether it is a keyword
                elif token_type == "IDENTIFIER":
                    if value in keywords:
                        counts["Keywords"] += 1
                        tokens.append(("KEYWORD", value))
                    else:
                        counts["Identifiers"] += 1
                        tokens.append(("IDENTIFIER", value))

                    counts["Total Tokens"] += 1
                    pos += len(value)
                    break

                # Number
                elif token_type == "NUMBER":
                    counts["Numbers"] += 1
                    counts["Total Tokens"] += 1
                    tokens.append(("NUMBER", value))
                    pos += len(value)
                    break

                # Operator
                elif token_type == "OPERATOR":
                    if value in operators_set:
                        counts["Operators"] += 1
                        counts["Total Tokens"] += 1
                        tokens.append(("OPERATOR", value))
                        pos += len(value)
                        break

        # No pattern matched
        if match is None:
            print(f"Warning: Unknown token '{source_code[pos]}' at position {pos}")
            pos += 1

    return counts, tokens
```

Enter a C Program

```python
source_code = """
int main()
{
    int count = 10;
    float value = 20.5;

    // This is a comment

    if (count >= 10)
    {
        count += 5;
        value = value + 2.5;
    }

    return 0;
}
"""
```

Perform Lexical Analysis

```python
counts, tokens = analyze_c(source_code)
```

Display Lexical Analysis Results

```python
print("Lexical Analysis Results:")

print("Keywords:", counts["Keywords"])
print("Identifiers:", counts["Identifiers"])
print("Operators:", counts["Operators"])
print("Numbers:", counts["Numbers"])
print("Total Tokens:", counts["Total Tokens"])
```

Display Recognized Tokens

```python
print("Recognized Tokens:\n")

for token_type, value in tokens:
    print(f"{token_type:12} -> {value}")
```

# **TASK 5 — Algorithm / Pseudocode**

# Algorithm

1. Start.
2. Read the C source program as input.
3. Initialize the token counts to zero.
4. Set `pos = 0` to point to the current character.
5. Try each regular-expression pattern in the specified order.
6. Ignore whitespace and comments.
7. If an identifier is found, check whether it belongs to the keyword set.
8. If a number is found, classify it as a number.
9. If an operator is found, classify it as an operator.
10. Increase the corresponding token count.
11. Move `pos` forward by the length of the recognized token.
12. If no pattern matches, report the character as an unknown token and move forward.
13. Repeat until the entire source program is scanned.
14. Display the number of keywords, identifiers, operators, numbers, and total tokens.
15. Stop.

# **Explanation of Working**

# Working of the Lexical Analyzer

The lexical analyzer scans the C source code from left to right using the variable `pos`.

At every position, the program tries the regular-expression patterns stored in `token_patterns_ordered`.

- Comments and whitespace are ignored.
- Numbers are identified using regular expressions.
- Identifiers are matched using `[A-Za-z_][A-Za-z0-9_]*`.
- After an identifier is found, it is compared with the `keywords` set. If it matches a keyword, it is classified as a keyword; otherwise, it is an identifier.
- Operators are recognized using regular expressions and the `operators_set`.
- Multi-character operators such as `==`, `+=`, `>=`, and `++` are checked before single-character operators.
- The variable `counts` maintains the number of each type of token.
- If no pattern matches a character, a warning is displayed and `pos` is advanced to avoid an infinite loop.

For example:

int count = 10;

is recognized as:

int   → Keyword
count → Identifier
=     → Operator
10    → Number
;     → Operator

# **Why Pattern Order Matters**

# Importance of Pattern Order

The order of regular-expression patterns is important because some tokens have common prefixes.

For example:

`==` contains `=`

Therefore, `==` must be checked before `=`.

Similarly:

`+=` must be checked before `+`

`++` must be checked before `+`

`>=` must be checked before `>`

Comments must also be checked before `/`, because `//` and `/*` begin with `/`.

Thus, longer and more specific patterns are placed before shorter and more general patterns.
